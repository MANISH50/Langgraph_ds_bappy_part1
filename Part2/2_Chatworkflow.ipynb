```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import Annotated, TypedDict
from pydantic import BaseModel, Field
from langchain_core.messages import BaseMessage, HumanMessage, AIMessage, SystemMessage
from dotenv import load_dotenv
from langgraph.checkpoint.memory import MemorySaver
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
model= ChatOpenAI(model_name="gpt-4", temperature=0.7, max_tokens=1000)
```

```python
from langgraph.graph.message import add_messages

class ChatState(TypedDict):
    messages: Annotated[list[BaseMessage], add_messages] 
    ##Here we are using reducer concept.
    #BaseMessage is a class that represents a message in the chat. It can be either a HumanMessage, AIMessage, or SystemMessage. The add_messages function is a reducer that adds messages to the state.
```

```python
def chat_node(state: ChatState):
    message = state["messages"]
    response = model.invoke(message)
    return {"messages": [response]}
```

```python
checkpoint= MemorySaver() #To save the entire state of the chatbot, including the messages and any other relevant information in RAM.
graph= StateGraph(ChatState)
#Add Nodes
graph.add_node('Chat_Node', chat_node)
#Add Edges
graph.add_edge(START, 'Chat_Node')
graph.add_edge('Chat_Node', END)
chatbot= graph.compile(checkpointer=checkpoint) #Compile the graph with the checkpoint to save the state of the chatbot.
```

```python
user_id = input("Enter user ID: ").strip()
print("user_id: ", user_id)
config = {"configurable": {"thread_id": user_id}}

while True:
    user_input = input("User: ")
    print(user_id,":", user_input)

    if user_input.lower() in ["exit", "quit"]:
        print("Exiting the chat.")
        break

    initial_state = {
        "messages": [HumanMessage(content=user_input)]
    }

    interact_messages = chatbot.invoke(
        initial_state,
        config=config
    )

    print("Bot:", interact_messages["messages"][-1].content)
```
