```python
from langgraph.graph import START,END,StateGraph
from langchain_openai import ChatOpenAI
from typing import TypedDict, Literal
from dotenv import load_dotenv
from pydantic import BaseModel, Field
```

```python
load_dotenv()
```

```python
model= ChatOpenAI(model_name="gpt-4o-mini", temperature=0.7)
```

```python
# model.invoke("Hey!").content
```

First Structured Output

```python
class SentimentSchema(BaseModel):
    sentiment: Literal["positive","negative"] = Field(description="The sentiment of the review")

class DiagnosisNotes(BaseModel):
    issue_type: Literal["UX","Bugs","Performance","Support","Other"] = Field(description="The type of issue identified in the diagnosis notes")
    tone: Literal["angry","calm", "excited", "frustrated","neutral","happy","satisfied"] = Field(description="The tone of the diagnosis notes")
    urgency: Literal["low","medium","high"] = Field(description="The urgency of the issue identified in the diagnosis notes")
```

```python
structured_sentiment = model.with_structured_output(SentimentSchema)
structured_diagnosis = model.with_structured_output(DiagnosisNotes)
```

```python
# prompt = "The"
# structured_sentiment.invoke(prompt)
```

```python
class ReviewState(TypedDict):
    review: str = Field(description="The review text")
    sentiment: Literal["positive","negative"] = Field(description="The sentiment of the review")
    diagnosis: dict = Field(description="The diagnosis notes for the review")
    response: str
```

```python
def find_sentiment(state:ReviewState):
    prompt = f"Please analyze the following review and determine if the sentiment is positive or negative:\n\n{state['review']}"
    sentiment= structured_sentiment.invoke(prompt)
    return {"sentiment": sentiment}
```

```python
def run_diagnosis(state:ReviewState):
    prompt = f"Please analyze the following review and provide diagnosis notes including issue type, tone, and urgency:\n\n{state['review']}"
    diagnosis= structured_diagnosis.invoke(prompt)
    return {"diagnosis": diagnosis.model_dump()}
```

```python
def positive_response(state:ReviewState):
    prompt=f"""Write a warm-positive response to the following review:\n\n{state['review']}
    And also ask user to leave a review on our website."""
    response= model.invoke(prompt)
    return {"response": response.model_dump()}

def negative_response(state:ReviewState):
    prompt=f"""Your'are a product support agent. Write an empathetic response to the following review:\n\n{state['review']}
    which has been diagnosed as {state['diagnosis']['issue_type']} issue with {state['diagnosis']['tone']} tone and {state['diagnosis']['urgency']} urgency."""
    
    response= model.invoke(prompt)
    return {"response": response.model_dump()}
```

```python
def check_sentiment(state: ReviewState) -> Literal["positive_response", "run_diagnosis"]:
    if state["sentiment"] == "positive":
        return "positive_response"
    else:
        return "run_diagnosis"      
```

```python
graph= StateGraph(ReviewState)

##Add Nodes
graph.add_node("find_sentiment",find_sentiment)
graph.add_node("run_diagnosis",run_diagnosis)
graph.add_node("positive_response",positive_response)
graph.add_node("negative_response",negative_response)

##Add Edges
graph.add_edge(START,"find_sentiment")
graph.add_conditional_edges("find_sentiment", check_sentiment)
graph.add_edge("positive_response",END)

##
graph.add_edge("run_diagnosis","negative_response")
graph.add_edge("negative_response",END)
graph.add_edge("find_sentiment","positive_response")
```

```python
workflow= graph.compile()
workflow
```

```python
user_input={
    "review": "I recently purchased your product and I am extremely satisfied with its performance. The user interface is intuitive and the features are exactly what I needed. Keep up the great work!"
}
workflow.invoke(user_input)
```
