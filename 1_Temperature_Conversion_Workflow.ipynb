Imports

```python
from langgraph.graph import  StateGraph,START,END
from typing import TypedDict
```

# Define State

```python
class TemperatureState(TypedDict):
    temperature_celsius: float
    temperature_fahrenheit: float
    label_weather: str
```

```python
#Convert temperature from Celsius to Fahrenheit
def convert_temp(state: TemperatureState) -> TemperatureState:
    celcius= state['temperature_celsius']
    fahrenheit= (celcius * 9/5) + 32
    state['temperature_fahrenheit']= round(fahrenheit,2)
    return state
```

```python
#Define the weather label function
def label_weather(state: TemperatureState) -> TemperatureState:
    temperature_f= state['temperature_fahrenheit']
    if temperature_f < 32:
        state['label_weather'] = "Freezing"
    elif 32 <= temperature_f < 50:
        state['label_weather'] = "Cold"
    elif 50 <= temperature_f < 70:
        state['label_weather'] = "Mild"
    elif 70 <= temperature_f < 90:
        state['label_weather'] = "Warm"
    else:
        state['label_weather'] = "Hot"
    return state 
```

# Define and Compile

```python
#Define the state graph for temperature conversion
graph= StateGraph(TemperatureState)

graph.add_node('convert_temp',convert_temp)
graph.add_node('label_weather',label_weather)

##Create edges for the state graph
graph.add_edge(START,'convert_temp')
graph.add_edge('convert_temp','label_weather')
graph.add_edge('label_weather',END)
```

```python
#Compile the state graph
workflow= graph.compile()
```

```python
#Execute the workflow with an initial state
initial_state={'temperature_celsius': 10.5}

final_state= workflow.invoke(initial_state)
```
