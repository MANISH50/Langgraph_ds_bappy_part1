```python
from langgraph.graph import StateGraph,START,END
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from typing import TypedDict, Annotated
from pydantic import BaseModel,Field
import operator #Reducer
```

```python
load_dotenv()
```

```python
model= ChatOpenAI(model="gpt-4o-mini")
```

```python
class EvolutionSchema(BaseModel):
    feedback: str = Field(description="Feedback for the evolution")
    score: int = Field(description="Score for the evolution upto 10",ge=0, le=10)
```

```python
#Now, this will gonna add with the model structured output. That means if it needs to generate the output it needs feedback and score.
structured_output = model.with_structured_output(EvolutionSchema)
```

```python
essay="""
# Food Adulteration

Food adulteration is a serious problem that affects the health and safety of people. It means adding cheap, harmful, or unwanted substances to food to increase its quantity, appearance, or profit. Common examples include mixing water in milk, adding artificial colours to spices, and using low-quality oils.

Food adulteration can cause many health problems, such as stomach disorders, allergies, food poisoning, and long-term diseases. It is especially dangerous for children and elderly people. To prevent it, consumers should buy food from trusted sources, check labels and expiry dates, and avoid suspiciously cheap products.

Strict government regulations and regular quality checks are also necessary to control food adulteration. Public awareness plays an important role in protecting people from unsafe food. Therefore, everyone—including consumers, food producers, and authorities—must work together to ensure that the food we eat is safe and healthy.
"""
```

```python
prompt=f"Evaluate the english language for the following essay and provide a feeback and assign a score out of 10: {essay}"
```

```python
model_response=structured_output.invoke(prompt)
```

```python
print(model_response)
```

```python
class EssayEvaluation(TypedDict):
    essay: str
    language_feedback: str
    analysis_feeback: str
    claritive_feedback: str
    overall_feedback: int
    score: Annotated[list[int], operator.add]
    average_score: float
```

```python
def language_feeback_node(state: EssayEvaluation):
    # This node will provide feedback on the language used in the essay
    prompt = f'Evaluate the language quality of the following essay and provide a feedback and assign a score out of 10 \n {state["essay"]}'
    model_response = structured_output.invoke(prompt)
    return {'language_feedback': model_response.feedback, 'score': [model_response.score]}
```

```python
def analysis_feeback_node(state: EssayEvaluation):
    # This node will provide feedback on the analysis of the essay
    prompt = f'Evaluate the depth of analysis of the following essay and provide a feedback and assign a score out of 10 \n {state["essay"]}'    
    model_response = structured_output.invoke(prompt)
    return {'analysis_feeback': model_response.feedback, 'score': [model_response.score]}
```

```python
def claritive_feedback(state: EssayEvaluation):

    prompt = f'Evaluate the clarity of thought of the following essay and provide a feedback and assign a score out of 10 \n {state["essay"]}'
    output = structured_output.invoke(prompt)

    return {'claritive_feedback': output.feedback, 'score': [output.score]}
```

```python
def final_evaluation(state: EssayEvaluation):

    # summary feedback
    prompt = f'Based on the following feedbacks create a summarized feedback \n language feedback - {state["language_feedback"]} \n depth of analysis feedback - {state["analysis_feeback"]} \n clarity of thought feedback - {state["claritive_feedback"]}'
    overall_feedback = model.invoke(prompt).content

    # avg calculate
    avg_score = sum(state['score'])/len(state['score'])

    return {'overall_feedback': overall_feedback, 'average_score': avg_score}
```

```python
graph= StateGraph(EssayEvaluation)

##Addimg nodes to the graph
graph.add_node("language_feedback", language_feeback_node)
graph.add_node("analysis_feedback", analysis_feeback_node)
graph.add_node("clarity_feedback", claritive_feedback)
graph.add_node("final_evaluation", final_evaluation)

##Adding the edges to the graph
graph.add_edge(START, "language_feedback")
graph.add_edge(START, "analysis_feedback")
graph.add_edge(START, "clarity_feedback")

graph.add_edge("language_feedback", "final_evaluation")
graph.add_edge("analysis_feedback", "final_evaluation")
graph.add_edge("clarity_feedback", "final_evaluation")

graph.add_edge("final_evaluation", END)

workflow= graph.compile()
```
